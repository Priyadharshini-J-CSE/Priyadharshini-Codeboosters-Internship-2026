```python
# Cell 2: Import Libraries
import pandas as pd
import chromadb

from sentence_transformers import SentenceTransformer
from groq import Groq
```

```python
# Cell 3: Create Sample College Notes Dataset
data = {
    "topic": [
        "Computer Networks",
        "DBMS",
        "Operating Systems",
        "Data Structures",
        "Machine Learning",
        "Cloud Computing",
        "Cyber Security",
        "Software Engineering"
    ],

    "content": [
        "OSI model contains 7 layers and TCP IP contains 4 layers.",
        "Normalization is used to reduce redundancy in databases.",
        "Round Robin is a CPU scheduling algorithm.",
        "Stack follows LIFO and Queue follows FIFO.",
        "Supervised learning uses labeled data.",
        "Cloud services include IaaS PaaS and SaaS.",
        "Encryption helps secure communication.",
        "SDLC consists of planning design coding testing and deployment."
    ]
}

df = pd.DataFrame(data)

df.to_csv("college_notes.csv", index=False)

df.head()
```

```python
# Cell 4: Load Dataset
df = pd.read_csv("college_notes(1).csv")

df
```

```python
# Cell 5: Load Embedding Model
embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

print("Embedding model loaded")
```

```python
# Cell 6: Create ChromaDB Client
chroma_client = chromadb.Client()

collection = chroma_client.create_collection(
    name="college_notes"
)

print("Collection Created")
```

```python
# Cell 7: Generate Embeddings and Store in ChromaDB
for idx, row in df.iterrows():

    embedding = embedding_model.encode(
        row["content"]
    ).tolist()

    collection.add(
        ids=[str(idx)],
        documents=[row["content"]],
        metadatas=[{"topic": row["topic"]}],
        embeddings=[embedding]
    )

print("Documents Stored Successfully")
```

# Cell 8: Configure Groq API

Get API Key from:

https://console.groq.com


```python
GROQ_API_KEY = "xxxxxxxxxxxxxx"

groq_client = Groq(
    api_key=GROQ_API_KEY
)
```

```python
# Cell 9: Retrieval Function
def retrieve_context(question, k=3):

    query_embedding = embedding_model.encode(
        question
    ).tolist()

    results = collection.query(
        query_embeddings=[query_embedding],
        n_results=k
    )

    docs = results["documents"][0]

    return "\n".join(docs)
```

```python
# Cell 10: RAG Pipeline
def ask_college_assistant(question):

    context = retrieve_context(question)

    prompt = f"""
You are a college knowledge assistant.

Answer only from the provided context.

Context:
{context}

Question:
{question}
"""

    response = groq_client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ],
        temperature=0
    )

    answer = response.choices[0].message.content

    return answer
```

```python
# Cell 11: Test the System
question = "What are the cloud service models?"

answer = ask_college_assistant(question)

print(answer)
```

```python
# Cell 12: Interactive Chatbot
while True:

    question = input("\nAsk Question : ")

    if question.lower() == "exit":
        print("Goodbye")
        break

    answer = ask_college_assistant(question)

    print("\nAnswer:")
    print(answer)
```
