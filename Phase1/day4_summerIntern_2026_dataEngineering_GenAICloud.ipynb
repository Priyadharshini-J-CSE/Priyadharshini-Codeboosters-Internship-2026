```python
#divide into x and y
#x-input feature and y-output feature
x=df[['Height(Inches)']]
y=df[['Weight(Pounds)']]
```

```python
from sklearn.model_selection import train_test_split
```

This line imports the train_test_split function from the model_selection module of the Python library scikit-learn.

Breakdown
sklearn → Short name for Scikit-learn
model_selection → Module used for splitting and selecting data for machine learning models
train_test_split → Function used to divide data into:
Training data → used to train the model
Testing data → used to test model accuracy

Why we use it

In Machine Learning, we should not train and test using the same data.

So we split the dataset like this:

80% → Training
20% → Testing
Example

variable1, variable2, variable3, variable4 = train_test_split(input_data, output_data, test_size=value)

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
x_train
```

```python
x_test
```

```python
#import the model -- linear regression
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression() #MODEL SELECTED
```

```python
#train the model by using the fit
model.fit(x_train,y_train)
```

```python
#test the model by using the predict
y_pred=model.predict(x_test)
```

```python
y_test
```

```python
#subtract the val to know whether it is correct or not - using metrices ( to check error and accuracy)
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score;
```

```python
mean_absolute_error(y_test,y_pred)
```

```python
mean_squared_error(y_test,y_pred)
```

```python
r2_score(y_test,y_pred)
#accuracy score : ans correctly 25%
```
